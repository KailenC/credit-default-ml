## Conclusion

Reweighting did not reduce the worst group gap (33.7 points). Four fewer false flags came with four more missed defaults. The paired marginal bootstrap interval for gap change includes zero. Probabilities overestimate risk on average, while most group calibration assessments lack sufficient bin support. These exploratory results do not establish fairness. The test set remains untouched.
